```python
from dotenv import load_dotenv|
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent 
```

```python
model= ChatGroq (model ="openai/gpt-oos-20b")
search= GoogleSerperAPIWrapper()
```

```python
agent=create_agent (
    model = model,
    tools= [search.run]
    system_prompt="you are the agents and can search for any question on google."
)
```

```python
question = "who won the woman's world cup final in 2025"
```

```python
response = agent.invoke = ({"messages": [{"role":"usere" , "content":question}]})
```

```python
print(response['messages'][-1].content)
```
