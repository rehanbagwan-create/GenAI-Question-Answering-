#####Prompts Templates - Dynamic Prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
out = StrOutputParser()
```

####My Self Chain Node 

```python
def transform_case(result:str):
    return result.upper
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompts = ChatPromptTemplate.from_messages([
    ("system", "You are a translator. Translate the following text into {language}."),
    ("human", "{query}")
])
```

```python
final_prompts = prompts.format_messages(language = "Hinglish ?", query = "I love  Python , and JavaScripts")
```

```python
res = llm.invoke(final_prompts)

print(res.content)
```

####Chains - Runnable

```python
chains = prompts | llm |out | transform_case
```
